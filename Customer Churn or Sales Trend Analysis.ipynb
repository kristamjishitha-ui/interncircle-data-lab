```python
import pandas as pd
import numpy as np
import random

print("Libraries imported successfully!")
```

```python
random.seed(42)

products = [
    ("Laptop", "Electronics", 65000),
    ("Smartphone", "Electronics", 28000),
    ("Headphones", "Electronics", 2500),
    ("Office Chair", "Furniture", 9000),
    ("Desk", "Furniture", 12000),
    ("Backpack", "Accessories", 1800),
    ("Running Shoes", "Footwear", 3500),
    ("T-Shirt", "Clothing", 900),
    ("Jeans", "Clothing", 1800),
    ("Coffee Maker", "Home Appliances", 5500)
]

segments = ["Consumer", "Corporate", "Small Business"]
regions = ["North", "South", "East", "West"]

rows = []

for i in range(1, 201):

    product, category, price = random.choice(products)

    quantity = random.randint(1, 5)
    discount = random.choice([0, 5, 10, 15])

    sales = price * quantity * (1 - discount / 100)

    profit = sales * random.uniform(0.10, 0.30)

    rows.append([
        f"ORD{i:04d}",
        pd.Timestamp("2025-01-01") + pd.Timedelta(days=random.randint(0, 364)),
        f"CUST{random.randint(1001, 1060)}",
        random.choice(segments),
        product,
        category,
        random.choice(regions),
        quantity,
        discount,
        round(sales, 2),
        round(profit, 2)
    ])

df = pd.DataFrame(rows, columns=[
    "Order_ID",
    "Order_Date",
    "Customer_ID",
    "Customer_Segment",
    "Product",
    "Category",
    "Region",
    "Quantity",
    "Discount_Percent",
    "Sales",
    "Profit"
])

print("E-commerce dataset created successfully!")

display(df.head())
```

```python
df["Year"] = df["Order_Date"].dt.year

df["Month"] = df["Order_Date"].dt.month

df["Month_Name"] = df["Order_Date"].dt.strftime("%B")

df["Year_Month"] = df["Order_Date"].dt.strftime("%Y-%m")

df["Profit_Margin"] = df["Profit"] / df["Sales"]

print("Feature engineering completed!")
```

```python
print("Dataset information:")
df.info()
```

```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
print("Missing values:")
display(df.isnull().sum())
```

```python
monthly_sales = df.groupby("Year_Month")["Sales"].sum()

print("Monthly Sales Trend:")
display(monthly_sales)
```

```python
segment_sales = df.groupby("Customer_Segment")["Sales"].sum()

print("Sales by Customer Segment:")
display(segment_sales.sort_values(ascending=False))
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print("Sales by Category:")
display(category_sales.sort_values(ascending=False))
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print("Sales by Region:")
display(region_sales.sort_values(ascending=False))
```

```python
top_products = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Products:")
display(top_products)
```

```python
top_customers = (
    df.groupby("Customer_ID")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Customers:")
display(top_customers)
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_quantity = df["Quantity"].sum()
average_order_value = total_sales / total_orders

print("----- BUSINESS SUMMARY -----")
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
print("Average Order Value:", round(average_order_value, 2))
```

```python
excel_file = "InternCircle_Task3_Ecommerce_Sales_Analysis.xlsx"

with pd.ExcelWriter(excel_file, engine="openpyxl") as writer:

    df.to_excel(
        writer,
        sheet_name="Sales_Data",
        index=False
    )

    monthly_sales.to_excel(
        writer,
        sheet_name="Monthly_Trend"
    )

    segment_sales.to_excel(
        writer,
        sheet_name="Customer_Segments"
    )

    category_sales.to_excel(
        writer,
        sheet_name="Categories"
    )

    region_sales.to_excel(
        writer,
        sheet_name="Regions"
    )

    top_products.to_excel(
        writer,
        sheet_name="Top_Products"
    )

    top_customers.to_excel(
        writer,
        sheet_name="Top_Customers"
    )

print("Excel file created successfully!")
```
